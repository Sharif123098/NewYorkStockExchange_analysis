### Reading CSV files

```python
import pandas as pd
```

```python
da = pd.read_csv("fundamentals.csv")
db = pd.read_csv("securities.csv")
dc = pd.read_csv("prices-split-adjusted.csv")
dd = pd.read_csv("prices.csv")
```

### Checking for unique values in each data file's column

```python
for column in da.columns:
    print(f"{column}: {da[column].nunique()}")
```

### Checking for null values in each data file's column

```python
for column in da.columns:
    print(f"{column}: {da[column].isnull().sum()}")
```

### Removing securities from securities whose fundamentals doesn't exists

```python
fundamentals = pd.read_csv("fundamentals.csv")
securities = pd.read_csv("securities.csv")

securities = securities[
    securities["Ticker symbol"].isin(fundamentals["Ticker Symbol"])
]


for column in securities.columns:
    print(f"{column}: {securities[column].nunique()}")
```

### Removing Securities from prices whose fundamentals doesn't exists

```python
prices = pd.read_csv("prices.csv")

prices = prices[
    prices["symbol"].isin(securities["Ticker symbol"])
]

for column in prices.columns:
    print(f"{column}: {prices[column].nunique()}")
```

### Removing Securities from price-adjusted whose fundamentals doesn't exists

```python
prices_split_adjusted = pd.read_csv("prices-split-adjusted.csv")

prices_split_adjusted = prices_split_adjusted[
    prices_split_adjusted["symbol"].isin(securities["Ticker symbol"])
]

for column in prices_split_adjusted.columns:
    print(f"{column}: {prices_split_adjusted[column].nunique()}")
```
